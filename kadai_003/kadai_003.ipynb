# NumPyで配列の操作や演算をしよう

行列AとBの積を計算して表示し、積の行列の要素の最大値を求めて表示する。

## Step1. NumPyをインポートする

```python
import numpy as np
```

## Step2. AとBの行列を多次元配列で書く

Aは3行2列、Bは2行4列。

```python
A = np.array([[0, 1],
              [2, 3],
              [4, 5]])
B = np.array([[0, 1, 2, 3],
              [4, 5, 6, 7]])
```

## Step3. 行列の積を計算する

Aの列の数（2）とBの行の数（2）が同じなので積を計算でき、結果は3行4列になる。

```python
C = np.dot(A, B)
```

## Step4. 計算結果を表示する

```python
print(C)
```

## Step5. 行列の要素の最大値を求めて表示する

```python
print(np.max(C))
```
