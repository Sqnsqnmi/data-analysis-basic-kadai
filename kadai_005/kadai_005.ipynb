# Matplotlibで、可視化した複数のグラフを1つの描画キャンパス上で同時に表示しよう

折れ線グラフ、sin関数、ヒストグラム、散布図を、1つの描画領域の中に2行3列で並べて表示する。

## Step1. matplotlib.pyplot と numpy をインポートする

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Step2〜4. 6個の領域を作り、4つのグラフを描いて表示する

`plt.subplots(2, 3)` で6個の領域（axes）を作り、`axes[行, 列]` で描く場所を指定する。
使うのは `axes[0, 0]`・`axes[0, 1]`・`axes[0, 2]`・`axes[1, 0]` の4つで、残りの2つは空けておく。

```python
# Step2. 2行3列で6個の領域を作る（tight_layout=True で領域が重ならないように調整する）
fig, axes = plt.subplots(2, 3, tight_layout=True, figsize=(12, 7))

# Step3. 各領域にグラフを表示する
# 折れ線グラフ
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0)
axes[0, 0].set_title('Line')

# sin関数
x1 = np.linspace(0, 10, 100)
y1 = 2 + 2 * np.sin(2 * x1)
axes[0, 1].plot(x1, y1, linewidth=2.0)
axes[0, 1].set(xlim=(0, 10), xticks=np.arange(0, 11),
               ylim=(-1, 5), yticks=np.arange(-1, 6))
axes[0, 1].set_title('sin')

# ヒストグラム（平均15、標準偏差5の正規分布から2000個の乱数）
x2 = np.random.normal(15, 5, 2000)
axes[0, 2].hist(x2, bins=30)
axes[0, 2].set_xlim(-10, 40)
axes[0, 2].set_title('Histogram')

# 散布図
np.random.seed(3)
x3 = 4 + np.random.normal(0, 2, 60)
y3 = 4 + np.random.normal(0, 2, len(x3))
axes[1, 0].scatter(x3, y3)
axes[1, 0].set_title('Scatter')

# Step4. グラフを表示する
plt.show()
```
